## Insight Generation and Report

## Insight 1 
According to the analysis, both hotels are being affected by a high rate of cancellations. The fact that a large portion of these reservations did not have a deposit indicates that customers are more inclined to cancel a reservation when they are not required to put down a deposit. The company should change the cancellation policy to make it a bit more restrictive. Additionally, it should conduct a brief survey of customers when they cancel a reservation to find out why they do so.


## Insight 2
It would also be advisable to conduct a satisfaction survey with clients who confirm their stay in order to find out their general impressions of the services offered. For example, according to the data, the bed and breakfast regime is the most requested, so it would be important to verify that its quality meets the expectations of the clients.


## Insight 3
On the other hand, there is a very low number of repeated clients, so a loyalty strategy could be created, especially for clients coming from Portugal, as they are the main clients for  hotels. In addition, their proximity   hotels would increase the likelihood of repeat stays

## Insight 4
Finally, the low occupancy of rooms by adults with children seems to indicate that none of the hotels has a family-friendly atmosphere. Consequently, it would be worth considering what services could be included to attract this type of client or whether, on the contrary, efforts should be focused on attracting couples and groups of adults to the hotel, offering more exclusive services

## insights 5
The hotel could increase direct bookings by offering exclusive website discounts, loyalty benefits, or special packages for customers who book directly.
